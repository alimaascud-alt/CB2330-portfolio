Dot product $\sum_{i=1}^{N} a_i b_i$ — multiplies corresponding values of two vectors and sums the product.

```python
def dot_product(a,b):
  total = 0
  for i in range(len(a)):
    product = a[i] * b[i]
    total += product

  return total
```

```python
import numpy as np

def dot_product_np(a,b):
  return np.dot(a,b)
```

```python
a = [2, 3, 4]
b = [5, 1, 2]

print(dot_product(a,b))
print(dot_product_np(a,b))
```
